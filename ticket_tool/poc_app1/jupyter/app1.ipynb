```python
# 必要なライブラリのインポート
import duckdb
import pandas as pd
import plotly.express as px
```

```python
# DuckDBに接続してサンプルデータを作成
con = duckdb.connect(':memory:')

# サンプルデータの作成
con.execute("""
    CREATE TABLE sales (
        date DATE,
        product VARCHAR,
        amount INTEGER
    )
""")

# データの挿入
con.execute("""
    INSERT INTO sales VALUES
    ('2024-01-01', 'Product A', 100),
    ('2024-01-02', 'Product A', 150),
    ('2024-01-03', 'Product A', 120),
    ('2024-01-01', 'Product B', 80),
    ('2024-01-02', 'Product B', 90),
    ('2024-01-03', 'Product B', 110)
""")
```

```python
# DuckDBからデータを読み出してPandasデータフレームに変換
df = con.execute("""
    SELECT 
        date,
        product,
        amount
    FROM sales
    ORDER BY date, product
""").df()
```

```python
# データの確認
print("データフレームの内容:")
display(df)
```

```python
# Plotlyを使用した可視化
fig = px.line(df, 
              x='date', 
              y='amount', 
              color='product',
              title='製品別売上推移')

# グラフの表示
fig.show()
```
